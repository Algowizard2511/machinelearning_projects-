```python
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
diabetes=load_diabetes()
```

```python
X=diabetes.data
Y=diabetes.target
```

```python
X.shape
```

```python
Y.shape
```

```python
features=diabetes.feature_names
features
```

```python
import pandas as pd
df=pd.DataFrame(X,columns=features,dtype=np.float64)
```

```python
df.info()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df[:],Y,random_state=2,test_size=0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.linear_model import Ridge
rg=Ridge(alpha=0.2,solver='cholesky')
rg.fit(X_train,y_train)
```

```python
y_pred=rg.predict(X_test)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
print('r2_score is : ',score)
print('coef is : ',rg.coef_)
print('intercept is : ',rg.intercept_)
```
